### CNEOS orbital elements space
Orbital elements data from backwards integration


```python
neos = pd.read_csv("../datos/sbdb_query_results_NEOS.csv")
neos["q"] = neos["a"]*(1-neos["e"])
neos = neos[neos['q'] < 1.30]

cneos_int = pd.read_csv("../datos/orbital_elements_integration.csv")
AU = 149597870 #km
deg = 180/np.pi

cneos_int["a"] = cneos_int["a (km)"]/AU
cneos_int["q"] = cneos_int["a"]*(1-cneos_int["e"])
cneos_int["i"] = cneos_int["i (rad)"]*deg
cneos_int['om'] = np.mod(cneos_int['Omega (rad)']*deg, 360)
cneos_int['w'] = np.mod(cneos_int['omega (rad)']*deg, 360)
cneos_int['ma'] = np.mod(cneos_int['M (rad)']*deg, 360)
cneos_int = cneos_int[cneos_int["q"] < 1.30]
```

### Estimating fireball size 
(assuming sphere and $\rho$ = 3000 - 8000 $kg$/$m^3$)

```python
joules = 4.184e+12
m = 1000
rho = 3000

chely_data
```

```python
I_energy = chely_data['Calculated Total Impact Energy (kt)'][indice]*joules
velocity = chely_data['Velocity (km/s)'][indice]*m
I_energy, velocity, size(I_energy, velocity, rho)
```

```python
merged['Impact Energy (J)'] = merged['Calculated Total Impact Energy (kt)']*joules
merged['Velocity (m/s)'] = merged['Velocity (km/s)']*m
merged['size (m)'] = size(merged['Impact Energy (J)'], merged['Velocity (m/s)'], rho)
```

```python
q_cneos = np.array(merged['q'])
e_cneos = np.array(merged['e'])
i_cneos = np.array(merged['i'])
size_cneos = np.array(merged['size (m)'])


fig = go.Figure()

# NEOS
fig.add_trace(go.Splom(
    dimensions=[
        dict(label='q (au)', values=q_cneos),
        dict(label='e', values=e_cneos),
        dict(label='i (deg)', values=i_cneos),
        dict(label='size (m)', values=size_cneos)
    ],
    name="Impactors",
    showupperhalf=False,
    diagonal_visible=False,
    marker=dict(color="#1f77b4", size=3, opacity=0.7)
))

# Layout minimalista
fig.update_layout(
    title_x=0.5,
    width=650,
    height=650,
    font=dict(family="Arial", color="#444444", size=14),
    plot_bgcolor='white',
    paper_bgcolor='white',
    dragmode='select',
    hovermode='closest',
    legend=dict(
        x=0.8, y=1.05,
        bgcolor='rgba(0,0,0,0)',
        borderwidth=0
    )
)

# Ajuste de ejes para estilo limpio
for axis in ['xaxis', 'xaxis2', 'xaxis3', 'yaxis', 'yaxis2', 'yaxis3']:
    fig.update_layout({axis: dict(
        showline=True,
        linewidth=1,
        linecolor="#333333",
        mirror=True,
        showgrid=True,
        gridcolor="#e6e6e6",
        zeroline=False
    )})

fig.show()
```
