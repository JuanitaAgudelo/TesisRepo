```python
import rebound as rb
import numpy as np
```

```python
rb.horizons.SSL_CONTEXT = 'unverified'

AU = 149597870 #km
day = 86400

sim = rb.Simulation()
sim.units = 'AU', 's', 'kg'

sim.add("Sun", hash="sun")
sim.add("399", hash="earth")
q=1
e=0.7
a = q/(1-e)
w=np.pi
sim.add(a=a, e=e, omega=w)


fig = rb.OrbitPlot(sim)
```
