```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
```

```python
neos = pd.read_csv("datos/sbdb_query_results_NEOS.csv")
neos = neos[neos['a'] < 2]

cneos_int = pd.read_csv("datos/orbital_elements_integration.csv")
AU = 149597870 #km
deg = 180/np.pi

cneos_int["a"] = cneos_int["a (km)"]/AU
cneos_int["i"] = cneos_int["i (rad)"]*deg
cneos_int['om'] = np.mod(cneos_int['Omega (rad)']*deg, 360)
cneos_int['w'] = np.mod(cneos_int['omega (rad)']*deg, 360)
cneos_int['ma'] = np.mod(cneos_int['M (rad)']*deg, 360)
cneos_int = cneos_int[cneos_int["a"] < 2]
```

```python
a_cneos = np.array(cneos_int['a'])
a_neos = np.array(neos['a'])
e_cneos = np.array(cneos_int['e'])
e_neos = np.array(neos['e'])
i_cneos = np.array(cneos_int['i'])
i_neos = np.array(neos['i'])


fig = go.Figure()

# NEOS
fig.add_trace(go.Splom(
    dimensions=[
        dict(label='a (au)', values=a_neos),
        dict(label='e', values=e_neos),
        dict(label='i (deg)', values=i_neos)
    ],
    name="NEOs",
    showupperhalf=False,
    diagonal_visible=False,
    marker=dict(color="#1f77b4", size=1, opacity=0.4)
))

# Impactadores
fig.add_trace(go.Splom(
    dimensions=[
        dict(label='a (au)', values=a_cneos),
        dict(label='e', values=e_cneos),
        dict(label='i (deg)', values=i_cneos)
    ],
    name="Earth Impactos",
    showupperhalf=False,
    diagonal_visible=False,
    marker=dict(color="#d62728", size=3, opacity=0.9)
))

# Layout minimalista
fig.update_layout(
    title="NEOS Orbital Elements Distribution and Earth Fireball Impactors",
    title_x=0.5,
    width=650,
    height=650,
    font=dict(family="Arial", color="#444444", size=14),
    plot_bgcolor='white',
    paper_bgcolor='white',
    dragmode='select',
    hovermode='closest',
    legend=dict(
        x=0.8, y=1.05,
        bgcolor='rgba(0,0,0,0)',
        borderwidth=0
    )
)

# Ajuste de ejes para estilo limpio
for axis in ['xaxis', 'xaxis2', 'xaxis3', 'yaxis', 'yaxis2', 'yaxis3']:
    fig.update_layout({axis: dict(
        showline=True,
        linewidth=1,
        linecolor="#333333",
        mirror=True,
        showgrid=True,
        gridcolor="#e6e6e6",
        zeroline=False
    )})

fig.show()
```
