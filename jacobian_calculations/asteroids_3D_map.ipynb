### TORUS PARAMETER CALCULATION FOR 95% VELOCITY CONTAINMENT

```python
# Load the state vectors data
state_vectors = np.loadtxt("NEOS_state_vectors.txt")

# Extract velocity components
vx = state_vectors[:, 3]
vy = state_vectors[:, 4] 
vz = state_vectors[:, 5]

print(f"Loaded {len(state_vectors)} state vectors")
print(f"Velocity components:")
print(f"  vx: min={vx.min():.4f} AU/yr, max={vx.max():.4f} AU/yr, mean={vx.mean():.4f} AU/yr, std={vx.std():.4f}")
print(f"  vy: min={vy.min():.4f} AU/yr, max={vy.max():.4f} AU/yr, mean={vy.mean():.4f} AU/yr, std={vy.std():.4f}")
print(f"  vz: min={vz.min():.4f} AU/yr, max={vz.max():.4f} AU/yr, mean={vz.mean():.4f} AU/yr, std={vz.std():.4f}")

# Calculate velocity magnitudes
velocity_magnitudes = np.sqrt(vx**2 + vy**2 + vz**2)
print(f"\nVelocity magnitude statistics:")
print(f"  Min: {velocity_magnitudes.min():.4f} AU/yr - {velocity_magnitudes.min()*(1/year)*AU_m*(1/1000):.4f} km/s")
print(f"  Max: {velocity_magnitudes.max():.4f} AU/yr - {velocity_magnitudes.max()*(1/year)*AU_m*(1/1000):.4f} km/s")
print(f"  Mean: {velocity_magnitudes.mean():.4f} AU/yr - {velocity_magnitudes.mean()*(1/year)*AU_m*(1/1000):.4f} km/s")
print(f"  Std: {velocity_magnitudes.std():.4f}")
print(f"  Median: {np.median(velocity_magnitudes):.4f}")
```

```python
# Calculate 95% confidence intervals for velocity magnitudes
percentile_2_5 = np.percentile(velocity_magnitudes, 2.5)
percentile_97_5 = np.percentile(velocity_magnitudes, 99.7)
percentile_95 = np.percentile(velocity_magnitudes, 97.5)

print(f"\n95% Confidence Intervals for Velocity Magnitudes:")
print(f"  2.5th percentile: {percentile_2_5:.4f}")
print(f"  97.5th percentile: {percentile_97_5:.4f}")
print(f"  95th percentile: {percentile_95:.4f}")
print(f"  Range (95%): {percentile_97_5 - percentile_2_5:.4f}")

# Calculate the distance from the z-axis for each velocity point
# This represents the distance from the central axis of the torus
r_xy = np.sqrt(vx**2 + vy**2)

print(f"\nDistance from z-axis (r_xy) statistics:")
print(f"  Min: {r_xy.min():.4f}")
print(f"  Max: {r_xy.max():.4f}")
print(f"  Mean: {r_xy.mean():.4f}")
print(f"  Std: {r_xy.std():.4f}")
print(f"  Median: {np.median(r_xy):.4f}")

# Calculate 95% percentiles for r_xy
r_xy_2_5 = np.percentile(r_xy, 2.5)
r_xy_97_5 = np.percentile(r_xy, 97.5)
r_xy_95 = np.percentile(r_xy, 95)

print(f"\n95% Confidence Intervals for r_xy:")
print(f"  2.5th percentile: {r_xy_2_5:.4f}")
print(f"  97.5th percentile: {r_xy_97_5:.4f}")
print(f"  95th percentile: {r_xy_95:.4f}")

# Calculate the z-component range for 95% of data
z_2_5 = np.percentile(np.abs(vz), 2.5)
z_97_5 = np.percentile(np.abs(vz), 97.5)
z_95 = np.percentile(np.abs(vz), 95)

print(f"\n95% Confidence Intervals for |vz|:")
print(f"  2.5th percentile: {z_2_5:.4f}")
print(f"  97.5th percentile: {z_97_5:.4f}")
print(f"  95th percentile: {z_95:.4f}")
```

```python
# =============================================================================
# TORUS PARAMETER CALCULATION
# =============================================================================

# Method 1: Based on velocity magnitude percentiles
# For a torus to contain 95% of points, we need to consider the 3D distribution
# The torus should have a major radius R and minor radius r such that 95% of points fall within

# Calculate the optimal torus parameters
# Major radius R should be the median distance from z-axis (central circle)
R_major = np.median(r_xy)
print(f"\n=== TORUS PARAMETER CALCULATION ===")
print(f"Method 1 - Based on r_xy distribution:")
print(f"  Major radius R (median r_xy): {R_major:.4f}")

# Minor radius r should be based on the z-component and the spread around the major radius
# We want to capture 95% of the data, so we use the 95th percentile
r_minor_method1 = np.percentile(np.sqrt((r_xy - R_major)**2 + vz**2), 95)
print(f"  Minor radius r (95th percentile distance from central circle): {r_minor_method1:.4f}")

# Calculate inner and outer radii
inner_radius_method1 = R_major - r_minor_method1
outer_radius_method1 = R_major + r_minor_method1

print(f"  Inner radius: {inner_radius_method1:.4f}")
print(f"  Outer radius: {outer_radius_method1:.4f}")

# Method 2: Based on velocity magnitude directly
# Use the 95th percentile of velocity magnitude as the outer boundary
max_velocity_95 = np.percentile(velocity_magnitudes, 97.5)
print(f"\nMethod 2 - Based on velocity magnitude:")
print(f"  95th percentile velocity magnitude: {max_velocity_95:.4f}")

# For a torus, we can approximate the major radius as the mean r_xy
R_major_method2 = np.mean(r_xy)
r_minor_method2 = max_velocity_95 / 2  # Approximate minor radius

inner_radius_method2 = R_major_method2 - r_minor_method2
outer_radius_method2 = R_major_method2 + r_minor_method2

print(f"  Major radius R: {R_major_method2:.4f}")
print(f"  Minor radius r: {r_minor_method2:.4f}")
print(f"  Inner radius: {inner_radius_method2:.4f}")
print(f"  Outer radius: {outer_radius_method2:.4f}")

# Method 3: Conservative approach using 97.5th percentiles
R_major_method3 = np.percentile(r_xy, 50)  # Median
r_minor_method3 = np.percentile(np.sqrt((r_xy - R_major_method3)**2 + vz**2), 99.7)

inner_radius_method3 = R_major_method3 - r_minor_method3
outer_radius_method3 = R_major_method3 + r_minor_method3

print(f"\nMethod 3 - Conservative (97.5th percentile):")
print(f"  Major radius R: {R_major_method3:.4f}")
print(f"  Minor radius r: {r_minor_method3:.4f}")
print(f"  Inner radius: {inner_radius_method3:.4f}")
print(f"  Outer radius: {outer_radius_method3:.4f}")
```

```python
# =============================================================================
# VERIFICATION AND VISUALIZATION
# =============================================================================

# Choose the best method (Method 1 is most statistically sound)
R_final = R_major
r_final = r_minor_method1
inner_radius_final = inner_radius_method1
outer_radius_final = outer_radius_method1

print(f"\n=== FINAL TORUS PARAMETERS ===")
print(f"Selected Method 1 (most statistically sound):")
print(f"  Major radius R: {R_final:.4f}")
print(f"  Minor radius r: {r_final:.4f}")
print(f"  Inner radius: {inner_radius_final:.4f}")
print(f"  Outer radius: {outer_radius_final:.4f}")

# Verify containment
def point_in_torus(x, y, z, R, r):
    """Check if a point (x,y,z) is within a torus with major radius R and minor radius r"""
    r_xy = np.sqrt(x**2 + y**2)
    distance_from_central_circle = np.sqrt((r_xy - R)**2 + z**2)
    return distance_from_central_circle <= r

# Check how many points are within the torus
points_in_torus = 0
for i in range(len(vx)):
    if point_in_torus(vx[i], vy[i], vz[i], R_final, r_final):
        points_in_torus += 1

containment_percentage = (points_in_torus / len(vx)) * 100

print(f"\n=== CONTAINMENT VERIFICATION ===")
print(f"Points within torus: {points_in_torus}/{len(vx)} ({containment_percentage:.2f}%)")

# If we need to adjust for exactly 95% containment
if containment_percentage < 95:
    # Find the 95th percentile distance from central circle
    distances_from_central = []
    for i in range(len(vx)):
        r_xy_i = np.sqrt(vx[i]**2 + vy[i]**2)
        dist = np.sqrt((r_xy_i - R_final)**2 + vz[i]**2)
        distances_from_central.append(dist)
    
    r_95_percentile = np.percentile(distances_from_central, 95)
    inner_radius_95 = R_final - r_95_percentile
    outer_radius_95 = R_final + r_95_percentile
    
    print(f"\nAdjusted for 95% containment:")
    print(f"  Minor radius r (95th percentile): {r_95_percentile:.4f}")
    print(f"  Inner radius: {inner_radius_95:.4f}")
    print(f"  Outer radius: {outer_radius_95:.4f}")
    
    # Verify the adjusted parameters
    points_in_torus_95 = 0
    for i in range(len(vx)):
        if point_in_torus(vx[i], vy[i], vz[i], R_final, r_95_percentile):
            points_in_torus_95 += 1
    
    containment_95 = (points_in_torus_95 / len(vx)) * 100
    print(f"  Adjusted containment: {points_in_torus_95}/{len(vx)} ({containment_95:.2f}%)")
    
    # Update final parameters
    R_final = R_final
    r_final = r_95_percentile
    inner_radius_final = inner_radius_95
    outer_radius_final = outer_radius_95
```

```python
# Create 3D visualization of velocity data with torus
fig = go.Figure()

# Add velocity points

fig.add_trace(go.Scatter3d(
    x=vx,
    y=vy,
    z=vz,
    mode='markers',
    marker=dict(
        size=2,
        color='blue',
        opacity=0.6
    ),
    name='Velocity Vectors (vx, vy, vz)',
    text=[f'vx={vx[i]:.4f}, vy={vy[i]:.4f}, vz={vz[i]:.4f}' for i in range(len(vx))],
    hovertemplate='%{text}<extra></extra>'
))

# Create torus surface for visualization
def create_torus_surface(R, r, n_theta=50, n_phi=50):
    """Create torus surface points for visualization"""
    theta = np.linspace(0, 2*np.pi, n_theta)
    phi = np.linspace(0, 2*np.pi, n_phi)
    theta, phi = np.meshgrid(theta, phi)
    
    x = (R + r * np.cos(phi)) * np.cos(theta)
    y = (R + r * np.cos(phi)) * np.sin(theta)
    z = r * np.sin(phi)
    
    return x, y, z

# Create torus surface
torus_x, torus_y, torus_z = create_torus_surface(R_final*1.4, r_final*2)

# Add torus surface

fig.add_trace(go.Surface(
    x=torus_x,
    y=torus_y,
    z=torus_z,
    opacity=0.3,
    colorscale='Reds',
    name='Torus Surface',
    showscale=False
))


# Update layout
fig.update_layout(
    title=f'Velocity Data with 95% Containment Torus<br>' +
          f'<sub>Major radius R={R_final:.4f}, Minor radius r={r_final:.4f}<br>' +
          f'Inner radius={inner_radius_final:.4f}, Outer radius={outer_radius_final:.4f}</sub>',
    scene=dict(
        xaxis_title='vx',
        yaxis_title='vy',
        zaxis_title='vz',
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.5)
        )
    ),
    width=1000,
    height=800
)

fig.show()

print(f"\n=== FINAL RESULT ===")
print(f"Torus parameters that contain 95% of velocity data:")
print(f"  Major radius R: {R_final:.4f}")
print(f"  Minor radius r: {r_final:.4f}")
print(f"  Inner radius: {inner_radius_final:.4f}")
print(f"  Outer radius: {outer_radius_final:.4f}")
print(f"  Actual containment: {containment_percentage:.2f}%")
```
