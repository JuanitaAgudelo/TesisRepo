```python
import numpy as np
import plotly.graph_objects as go

def cubos_en_esfera(R, a):
    """
    Divide una esfera de radio R en cubos de lado a (a << 1) y devuelve las coordenadas
    de los centros de los cubos que están completamente dentro de la esfera.

    Parámetros:
    R -- radio de la esfera
    a -- lado del cubo

    Retorna:
    centros -- np.ndarray de forma (N, 3) con las coordenadas de los centros de los cubos
    """
    # Definir los límites del cubo que contiene la esfera
    n = int(np.ceil(2*R/a))
    # Generar los centros de los cubos
    x = np.linspace(-R + a/2, R - a/2, n)
    y = np.linspace(-R + a/2, R - a/2, n)
    z = np.linspace(-R + a/2, R - a/2, n)
    X, Y, Z = np.meshgrid(x, y, z, indexing='ij')
    centros = np.vstack([X.ravel(), Y.ravel(), Z.ravel()]).T
    # Filtrar los centros que están dentro de la esfera
    dist = np.linalg.norm(centros, axis=1)
    dentro = dist <= (R - a/2)
    return centros[dentro]

def cubo_edges(center, a):
    """
    Devuelve las líneas (pares de puntos) que forman los bordes de un cubo centrado en 'center' y de lado 'a'.
    """
    x0, y0, z0 = center
    d = a / 2
    # 8 vértices del cubo
    vertices = np.array([
        [x0 - d, y0 - d, z0 - d],
        [x0 - d, y0 - d, z0 + d],
        [x0 - d, y0 + d, z0 - d],
        [x0 - d, y0 + d, z0 + d],
        [x0 + d, y0 - d, z0 - d],
        [x0 + d, y0 - d, z0 + d],
        [x0 + d, y0 + d, z0 - d],
        [x0 + d, y0 + d, z0 + d],
    ])
    # pares de índices de vértices que forman los 12 bordes
    edges = [
        (0,1), (0,2), (0,4),
        (1,3), (1,5),
        (2,3), (2,6),
        (3,7),
        (4,5), (4,6),
        (5,7),
        (6,7)
    ]
    return [(vertices[i], vertices[j]) for i,j in edges]

# Parámetros de ejemplo
R = 1.0
a = 0.3  # aumentado a 0.3
centros = cubos_en_esfera(R, a)

# Graficar los cubos (como líneas de los bordes)
fig = go.Figure()

# Opcional: también puedes mostrar los centros como puntos
fig.add_trace(go.Scatter3d(
    x=centros[:,0], y=centros[:,1], z=centros[:,2],
    mode='markers',
    marker=dict(size=3, color='blue', opacity=0.5),
    name='Centros de cubos'
))

# Dibujar los bordes de cada cubo
for centro in centros:
    for v0, v1 in cubo_edges(centro, a):
        fig.add_trace(go.Scatter3d(
            x=[v0[0], v1[0]],
            y=[v0[1], v1[1]],
            z=[v0[2], v1[2]],
            mode='lines',
            line=dict(color='black', width=2),
            showlegend=False
        ))

# Graficar la esfera
u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 50)
x_s = R * np.outer(np.cos(u), np.sin(v))
y_s = R * np.outer(np.sin(u), np.sin(v))
z_s = R * np.outer(np.ones_like(u), np.cos(v))

fig.add_trace(go.Surface(
    x=x_s, y=y_s, z=z_s,
    opacity=0.2,
    colorscale='Reds',
    showscale=False,
    name='Esfera'
))

fig.update_layout(
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='data'
    ),
    title='Cubos dentro de una esfera (con bordes)'
)

fig.show()

```

```python
import numpy as np
from scipy.integrate import nquad

# Definir la función en coordenadas cartesianas: f(x, y, z)
# Por ejemplo, f(x, y, z) = 1 (integral del volumen)
def f_cartesiana(x, y, z):
    # Aquí puedes cambiar la función a integrar
    return (x**2 + y**2 + z**2)**0.5  # Ejemplo: función constante
    #return 1
# Radio de la esfera
R = 2.0  # Cambia esto según tu esfera

# Límites de integración para x, y, z dentro de la esfera de radio R
def limites_y(x):
    return [-np.sqrt(R**2 - x**2), np.sqrt(R**2 - x**2)]

def limites_z(x, y):
    return [-np.sqrt(R**2 - x**2 - y**2), np.sqrt(R**2 - x**2 - y**2)]

# Realizar la integración numérica en coordenadas cartesianas
resultado, error = nquad(
    f_cartesiana,
    [
        limites_z,      # z: depende de x, y
        limites_y,      # y: depende de x
        [-R, R]         # x: de -R a R
    ]
)

print(f"Integral numérica en el volumen de la esfera (cartesianas): {resultado:.8f} (error estimado: {error:.2e})")
```
