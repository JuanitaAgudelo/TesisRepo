```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

### Integracion en hipercubo con centro en c: (1, 0, 0, 0, 6.284, 0) y tamano delta: (0.5, 0.5, 0.5, 1.0547, 1.0547, 1.0547) 

```python
df = pd.read_csv("count_objects.csv")
df['ratio'] = df['numerical_count']/df['theoretical_count']
df
```

```python
df_uniform = df[df['transformation_type'] == 'uniform_aewM']
fig = px.line(df_uniform, x=' iteration', y='ratio', color='objects_generated', markers=True)
fig.update_layout(title=dict(text="Trasformation (a,e,w,M)", font=dict(size=20)))
fig.show()

```

```python
df_complete = df[df['transformation_type'] == 'uniform_complete']
fig = px.line(df_complete, x=' iteration', y='ratio', color='objects_generated', markers=True)
fig.update_layout(title=dict(text="Complete Trasformation", font=dict(size=20)))
fig.show()
```

### Integracion en hipercubo con centro en c: (1, 0, 0, 0, 6.284, 0) y tamano delta: (0.2, 0.2, 0.2, 1.0547, 1.0547, 1.0547) 

```python
df2 = pd.read_csv("count_objects2.csv")
df2
```

```python
df_complete2 = df2[df2['transformation_type'] == 'uniform_complete']
df_complete2['ratio'] = df_complete2['numerical_count']/df_complete2['theoretical_count']
df_complete2
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Gráfico base
fig = px.line(
    df_complete2, 
    x='iteration', 
    y='ratio', 
    color='objects_generated', 
    markers=True
)

# Obtener los colores del gráfico
color_map = {}
for trace in fig.data:
    # A veces trace.name no es exactamente igual al valor de la columna, por eso lo forzamos a string
    color_map[str(trace.name)] = trace.line.color

# Agregar línea horizontal del promedio para cada categoría
for category in df_complete2['objects_generated'].unique():
    color = color_map.get(str(category), 'gray')  # color por defecto si no se encuentra
    avg_value = df_complete2.loc[df_complete2['objects_generated'] == category, 'ratio'].mean()

    fig.add_hline(
        y=avg_value,
        line=dict(color=color, width=2, dash='dot'),
        opacity=0.5
    )

# Ajustes visuales
fig.update_layout(
    title=dict(text="Complete Transformation", font=dict(size=20)),
    yaxis_title="Ratio",
    xaxis_title="Iteration",
    showlegend=True
)

fig.show()
```
