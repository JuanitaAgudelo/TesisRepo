```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

### Integracion en hipercubo con centro en c: (1, 0, 0, 0, 6.284, 0) y tamano delta: (0.5, 0.5, 0.5, 1.0547, 1.0547, 1.0547) 

```python
df = pd.read_csv("count_objects.csv")
df['ratio'] = df['numerical_count']/df['theoretical_count']
df
```

```python
df_uniform = df[df['transformation_type'] == 'uniform_aewM']

# Gráfico base
fig = px.line(
    df_uniform, 
    x=' iteration', 
    y='ratio', 
    color='objects_generated', 
    markers=True
)

# Obtener los colores del gráfico
color_map = {}
for trace in fig.data:
    # A veces trace.name no es exactamente igual al valor de la columna, por eso lo forzamos a string
    color_map[str(trace.name)] = trace.line.color

# Agregar línea horizontal del promedio para cada categoría
for category in df_uniform['objects_generated'].unique():
    color = color_map.get(str(category), 'gray')  # color por defecto si no se encuentra
    avg_value = df_uniform.loc[df_uniform['objects_generated'] == category, 'ratio'].mean()

    fig.add_hline(
        y=avg_value,
        line=dict(color=color, width=2, dash='dot'),
        opacity=0.5
    )

# Ajustes visuales
fig.update_layout(
    title=dict(text="Numeric Experiment P_xyvxvy Uniform dxyz = 0.5", font=dict(size=20)),
    yaxis_title="Ratio",
    xaxis_title="Iteration",
    showlegend=True
)

fig.show()

```
