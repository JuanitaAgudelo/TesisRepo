```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("count_objects.csv")
df['ratio'] = df['theoretical_count']/df['numerical_count']
df
```

```python
df_uniform = df[df['transformation_type'] == 'uniform_aewM']
fig = px.line(df_uniform, x=' iteration', y='ratio', color='objects_generated', markers=True)
fig.show()
```

```python
df_complete = df[df['transformation_type'] == 'uniform_complete']
fig = px.line(df_complete, x=' iteration', y='ratio', color='objects_generated', markers=True)
fig.show()
```
