\begin{equation}

    \partial_ay & \partial_ey & \partial_iy & \partial_\Omega y & \partial_wy & \partial_My \\

    \partial_av_z & \partial_ev_z & \partial_iv_z & \partial_\Omega v_z & \partial_wv_z & \partial_Mv_z
    \end{pmatrix}
\end{equation}
